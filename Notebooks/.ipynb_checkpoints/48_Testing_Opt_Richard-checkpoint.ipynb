# Data

```python
AS209 = {'inc': 34.97, 'pa': 85.76, 'dra':1.9e-3, 'ddec':-2.5e-3, 'rout': 1.2*2, 'disk_name': 'AS209'}
IMLup = {'inc': 47.5, 'pa': 144.5, 'dra':-1.5e-3, 'ddec':1e-3, 'rout': 1.71*2, 'disk_name': 'IMLup'}
Elias27 = {'inc': 56.2, 'pa': 118.8, 'dra':-5e-3, 'ddec':-8e-3, 'rout': 1.88*2, 'disk_name': 'Elias27'}
HD163296 = {'inc':46.7, 'pa':133.33, 'dra':-2.8e-3, 'ddec':7.7e-3, 'rout':1.7*2, 'disk_name': 'HD163296' }
Elias24 = {'inc': 29, 'pa': 45.7, 'dra':110.8e-3, 'ddec':-386.8e-3, 'rout': 1.03*2, 'disk_name': 'Elias24'}
HD143006 = {'inc': 18.6, 'pa': 169, 'dra':-5.9e-3, 'ddec':21.7e-3, 'rout': 0.518*3, 'disk_name': 'HD143006'}
```

```python
data = AS209

dir = "../uvtables/"
disk_name = data['disk_name']
data_file = dir +"uvtable_" + disk_name + "_continuum.npz"

# Huang 2018 
inc = data['inc']
pa = data['pa']
dra = data['dra']
ddec = data['ddec']
rout = data['rout']
```

```python
# Frank Parameters
N = 300
# load data
file = np.load(data_file)
try:
    u, v, Re, Im, Weights = file['u'], file['v'], file['Re'], file['Im'], file['w']
    Vis = Re + Im*1j
except KeyError:
    u, v, Vis, Weights = file['u'], file['v'], file['V'], file['weights']
```

# F2D

### Run

```python
geom = Geometry(inc, pa, dra, ddec)
frank2d = Frank2D(N, rout, geom)
```

```python
frank2d.process_vis(u, v, Vis, Weights)
```

```python
start_time = time.time()

kernel_params =  {'m': -2.35, 'c': 10**10.2, 'l': 7e4}
frank2d.fit(kernel_params = kernel_params, rtol = 1e-8 )

end_time = time.time()
print(f"F2D model in {(end_time - start_time)/60:.2f} minutes.")
```

```python
from frank.utilities import UVDataBinner
data =  frank2d.gridded_data
u_, v_, Vis_, Weights_ = data["u"], data["v"], data["vis"], data["weights"]

# Let's review the power spectrum.
baselines = np.hypot(u_, v_)         
cs, ms = ['#a4a4a4', 'k'], ['.', 'x']
bin_widths=[1e3, 1e5]

binned_vis = UVDataBinner(baselines, Vis_, Weights_, bin_widths[0])
logx = np.log10(binned_vis.uv)
logy = np.log10(np.abs(binned_vis.V)**2)

binned_vis2 = UVDataBinner(baselines, Vis_, Weights_, bin_widths[1])
logx2 = np.log10(binned_vis2.uv)
logy2 = np.log10(np.abs(binned_vis2.V)**2)

def linear_power_spectrum(logx, m, c):
    return m*logx + np.log10(c)
    
```

```python
plt.figure(figsize=(7, 3))
plt.plot(logx, logy, c=cs[0], marker=ms[0], ls='None', label=r'Obs., {:.0f} k$\lambda$ bins'.format(bin_widths[0]/1e3))
plt.plot(logx2, logy2, c=cs[1], marker=ms[1], ls='None', label=r'Obs., {:.0f} k$\lambda$ bins'.format(bin_widths[1]/1e3))


lgx = np.log10(np.geomspace(np.sort(baselines)[1], baselines.max(), 1000))

plt.plot(lgx, 8.2 - 2.0*lgx, ls='-')

plt.xlabel(r'log Baseline [$\lambda$]', size = 10)
plt.ylabel(r'log $|Vis_{Obs}|^{2}$ [Jy]', size = 10)
plt.legend(loc = 'best', fontsize = 'x-small')
plt.title("Power spectrum")
plt.ylim(-9, 0)
plt.xlim(5, 6.5)
plt.show()
```

Let's plot the posterior for different m values for a few choices of p at uv=10^5

```python
import time
start_time = time.time()
N_f1d = 50
FF = FourierBesselFitter(rout, N_f1d, geom)
sol = FF.fit(u_, v_, Vis_, Weights_)
total_time = time.time() - start_time
print("--- %s minutes ---", total_time/60, " minutes.")

GM = FF._GaussianModel
FT = GM._DFT
```

```python
p5 = [-2, -1, 0]
ms = np.linspace(-2, -5, 301)
logl = 4

minus_log_likelihood_array = []
intensity_array = []
logdetS_array = []
logdetD_array = []
jTDj_array = []

for pi in p5:
    minus_log_likelihood_array.append([])
    intensity_array.append([])
    logdetS_array.append([])
    logdetD_array.append([])
    jTDj_array.append([])
    for mi in ms:
        c5 = pi - 5*mi

        minus_log_likelihood = GM.m_log_posterior([mi, c5, logl])
        logdetS, logdetD, jTDj = GM.logdetS, GM.logdetD, GM.jTDj
        minus_log_likelihood_array[-1].append(minus_log_likelihood)
        logdetS_array[-1].append(logdetS)
        logdetD_array[-1].append(logdetD)
        jTDj_array[-1].append(jTDj)
        intensity_array[-1].append(GM.sol)
        print(f"---------------------------- (m, log_c):({mi}, {c5}) with likelihood: {minus_log_likelihood: 2e}")

```

```python
def plot_change(arr, **kwargs):
    plt.plot(ms, np.array(arr)-np.mean(arr), **kwargs)

minus_log_likelihood_array = np.array(minus_log_likelihood_array)
logdetS_array = np.array(logdetS_array)
logdetD_array = np.array(logdetD_array)
jTDj_array = np.array(jTDj_array)

for i, ls in zip(range(3), ['-', ':', '--']):
    plt.figure()
    plt.title(f'p(10^5) = 10^{p5[i]}')
    plot_change(minus_log_likelihood_array[i], label='Tot', c='C0', ls=ls)
    plot_change(logdetS_array[i], label='det(S)', c='C1', ls=ls)
    plot_change(logdetD_array[i], label='det(D)', c='C2', ls=ls)
    plot_change(jTDj_array[i], label='jDj', c='C3', ls=ls)
    if i == 0:
        plt.legend()

    plt.xlabel('slope')
    plt.ylabel('Change in quantity')

print(ms[minus_log_likelihood_array.argmin(axis=1)])
print(minus_log_likelihood_array[np.arange(3),minus_log_likelihood_array.argmin(axis=1)]-minus_log_likelihood_array.min())
```

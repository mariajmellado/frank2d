```python
# Let's review the gridded visibilities first.
baselines = np.hypot(u, v)         
grid = np.logspace(np.log10(min(baselines.min(), baselines[0])),
                           np.log10(max(baselines.max(), baselines[-1])),
                           10**4)

cs, ms = ['#a4a4a4', 'k'], ['.', 'x']
bin_widths=[1e3, 1e5]
binned_vis = UVDataBinner(baselines, Vis, np.ones_like(Weights), bin_widths[0])

x = binned_vis.uv
y = np.abs(binned_vis.V)
```

```python
weights = np.exp(logp - np.max(logp))  # normaliza para evitar overflow

# regresión ponderada m -> logc
A = np.vstack([m, np.ones_like(m)]).T
a, b = np.linalg.lstsq(A * np.sqrt(weights[:,None]), logc * np.sqrt(weights), rcond=None)[0]

print("Pendiente estimada a =", a, " intercepto b =", b)
```

```python
u_raw = logc - (a*m + b)
v_raw = m
su = np.std(u_raw)
sv = np.std(v_raw)
```

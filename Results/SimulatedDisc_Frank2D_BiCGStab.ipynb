```python
import numpy as np
import matplotlib.pyplot as plt
import time
```

```python
# frank2d
from frank2d import Frank2D
from frank2d.geometry import Geometry
from frank2d.plot import Plot
from frank2d.posterior_optimization import *
from frank2d.constants import rad_to_arcsec, deg_to_rad
from frank2d.posterior_optimization import MAPEstimator
```

```python
dir = "./../../data/"
data_file = dir + "uvtable_nonsym_disc_blob_noisy.npz"
inc = 0
pa = 0
dra = 0
ddec = 0
rout = 1*2

# Load data
file = np.load(data_file)
u, v, Vis, Weights = file['u'], file['v'], file['vis'], file['weights']
uvtable = {'u': u, 'v': v, 'vis': Vis, 'weights': Weights }
```

```python
geom_f2d = Geometry(inc, pa, dra, ddec)
```

## F2D

```python
N = 300
```

```python
param_N80 = {'m': -4.572442130254884, 'c': 1.0084384650978846e+24, 'l': 68720.771036978}
```

## Performing with different rtols
This experiment makes sense when the tolerance for Ax = b is higher.

```python
rtols = [1e-6, 1e-7, 1e-8, 1e-9, 1e-10, 1e-11, 1e-12]
```

```python
times_rtol = []
```

```python
for i in rtols:
    start_time = time.time()
    frank2d = Frank2D(N, rout)
    frank2d.fit(uvtable, kernel_params = param_N80, rtol=i)
    Plot(frank2d, geom_f2d).cg_tolerance()
    end_time = time.time()
    times_rtol.append(end_time - start_time)
```

```python
iterations = [135, 178, 219, 234, 280, 373, 657]
converged = np.array([False, False, True, True, True, True, True])
```

```python
marker_mask_times = np.where(converged, times_rtol, np.nan)
marker_mask_iter = np.where(converged, iterations, np.nan)
```

```python
plt.figure(figsize=(10,3))
plt.plot(rtols, times_rtol, marker='x', ls=':')
plt.yscale('log')
plt.xscale('log')
plt.ylabel(' Time (s)')
plt.plot(rtols, marker_mask, 
         marker='o',
         linestyle='',
         color='purple',
         markersize=8, 
         label='Converged)')
plt.legend()
plt.xlabel(' Conjugate Gradient tolerance')


plt.figure(figsize=(10,3))
plt.plot(rtols, iterations, marker='x', ls=':')
plt.yscale('log')
plt.xscale('log')
plt.plot(rtols, marker_mask_iter, 
         marker='o',
         linestyle='',
         color='purple',
         markersize=8, 
         label='Converged)')
plt.ylabel('Max CG iterations')
plt.xlabel(' Conjugate Gradient tolerance')

```
